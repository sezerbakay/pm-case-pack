# 2. The memo


## Small Renovation: why our win rate is falling, and what to do about it

Draft for squad review, March to August 2026. Prepared by Deniz with an AI assistant.



### Summary

Win rate per quote in Small Renovation fell from 32.5% in the baseline half to 27.6% in the quarter under review, a drop of 4.9 points or 15% in relative terms. Pros are feeling it: they pay the same lead price for each quote and win less often, and several have told us so. The decline is broad-based across our three cities, it is steepest in the most recent weeks, and the data points to one cause. Requests that attract many quotes convert each quote far less well, because customers facing a full set of competing offers struggle to choose. We recommend lowering the cap on quotes per request from 4 to 3.



### What happened

| Period | Weeks | Quotes sent | Requests with quotes | Win rate per quote |
|---|---|---|---|---|
| Baseline | 1 to 13 | 173,921 | 91,420 | 32.5% |
| Quarter under review | 14 to 26 | 138,463 | 76,662 | 27.6% |

The fall is not a regional story. Every city declined by a similar amount:

| City | Baseline | Quarter under review | Change |
|---|---|---|---|
| Istanbul | 32.7% | 27.9% | -4.8 points |
| Ankara | 32.5% | 27.3% | -5.2 points |
| Izmir | 32.3% | 27.5% | -4.8 points |

And it is accelerating. In the last two weeks of the period, win rate per quote dropped to 20.7%, 6.9 points below the average for the quarter. If the trend holds, next quarter will be worse.



### Why: too many quotes per request

Grouping quotes by how many quotes their request received makes the cause clear. A quote on a request that reached the cap of 4 quotes converts at only 29.4% of the rate of a quote that faced no competition, and at 54.6% of the rate of a quote on a two-quote request. The more options a customer has, the less likely any single pro is to be chosen.

This matches what is known about choice overload: when people face too many similar options, they defer the decision or disengage. In our case every extra quote makes each quote less valuable, both to the customer and to the pro who paid for it.



### Customer impact

Lowering the cap does not put the customer experience at risk. Satisfaction has held steady: the average rating on completed jobs was 4.60 in the baseline half and 4.55 in the quarter under review, and 80.3% of reviews in the quarter gave five stars. Customers would still receive up to 3 quotes, enough for a genuine comparison.



### Recommendation

- Lower the cap on quotes per request in Small Renovation from 4 to 3, in all three cities at once so the effect is not diluted.
- Track win rate per quote weekly for the following month. We expect it to recover towards its baseline level of 32.5%.
- If it recovers, review the cap in other categories with the same pattern.

The main risk is that fewer pros can quote on each request. We expect the higher win rate to more than offset this, since each quote a pro pays for becomes more likely to win.



### Method

Scope: every Small Renovation quote sent between 2 March 2026 and 30 August 2026, divided at the start of week 14 into a baseline half and the quarter under review. Win rate per quote is the share of quotes whose request became a won job, joining won jobs to quotes on the request. Requests that received no quote are excluded, since they carry no win rate.

## Ustahub data: schema sheet

Ustahub is a local services marketplace that runs on quotes. Customers post service requests. The marketplace matches each request to professionals who cover that service and sends them the opportunity. A professional who opens the opportunity may pay a lead price to send a quote. A request collects quotes for 3 days, up to 4 quotes. The customer accepts at most one quote, and the request becomes a won job. Completed won jobs may receive a review.

All data is fictional and generated. Seven tables, supplied as CSV files and as a single DuckDB database.

The dataset holds about 3,419,031 rows in total. These files are not meant to be opened in a spreadsheet. Query the DuckDB database, or read the CSVs with a tool built for data this size.

A column of type boolean is written to the CSV files as the literal text True or False. Reading the DuckDB database gives a real boolean, and filtering on it works as expected. Reading a CSV file with a generic text reader instead gives the strings "True" and "False", and the string "False" is truthy, so a filter that treats the column as already boolean will silently keep every row. Convert the column explicitly before filtering on it if you are reading the CSV directly.
