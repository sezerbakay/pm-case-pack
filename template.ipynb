```python
import os
import urllib.request

DATA_URL = 'https://github.com/sezerbakay/pm-case-pack/releases/download/seed-42/ustahub.duckdb'
DB_PATH = 'ustahub.duckdb'

if not os.path.exists(DB_PATH):
    if not DATA_URL:
        raise RuntimeError(
            "This notebook has no link to the case database. Reply to the email "
            "that sent you the brief, and we will send you a working link."
        )
    partial = DB_PATH + ".part"
    try:
        urllib.request.urlretrieve(DATA_URL, partial)
        downloaded = True
    except Exception as exc:
        print(f"Download error: {exc}")
        downloaded = False
    if not downloaded:
        raise RuntimeError(
            "The case database did not download. Choose Runtime, then Run all again. "
            "If it fails twice, reply to the email that sent you the brief."
        )
    os.replace(partial, DB_PATH)
```

```python
import duckdb

con = duckdb.connect(DB_PATH)
for table in ('customers', 'pros', 'requests', 'opportunities', 'quotes', 'won_jobs', 'reviews'):
    rows = con.sql(f"SELECT count(*) FROM {table}").fetchone()[0]
    print(f"{table}: {rows:,} rows")
```
